# Karhutla Indonesia & Potensi Permintaan Masker

Analisis eksploratif. Data penjualan demo bersifat simulasi. Ganti dengan data transaksi berizin untuk insight bisnis nyata.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
fires = pd.read_csv(ROOT / 'data/processed/monthly_hotspots.csv', parse_dates=['month'])
fires.head()
```

```python
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(fires['month'], fires['hotspot_count'])
ax.set(title='Tren Deteksi Hotspot Bulanan Indonesia', xlabel='Bulan', ylabel='Jumlah deteksi')
plt.show()
```

## Interpretasi
- Identifikasi bulan puncak dan pola musiman.
- Cocokkan periode dengan PM2.5 dan penjualan aktual jika tersedia.
- Jangan menyimpulkan sebab-akibat dari korelasi sederhana.
